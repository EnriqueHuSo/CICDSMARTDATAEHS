# 5.Grants – Seguridad de acceso Medallion

Este notebook aplica los permisos acordados para los grupos sincronizados con Unity Catalog.

**Catálogo:** `dbw_proyecto_final_ehs`  
**Esquemas:** `bronze`, `silver`, `golden`

> Ejecutar con una identidad que tenga autoridad para administrar los privilegios del catálogo y los esquemas. No crea ni modifica usuarios o grupos.

## 1. Matriz de permisos

| Grupo | Bronze | Silver | Golden |
|---|---|---|---|
| `grp_data_engineers` | SELECT | SELECT, MODIFY, CREATE TABLE | SELECT, MODIFY, CREATE TABLE |
| `grp_data_analysts` | Sin acceso | Sin acceso | SELECT |
| `grp_powerbi_readers` | Sin acceso | Sin acceso | SELECT |

Todos reciben `USE CATALOG`. `USE SCHEMA` se concede solo en los esquemas autorizados.

## 2. Data Engineers

```sql
GRANT USE CATALOG ON CATALOG dbw_proyecto_final_ehs TO `grp_data_engineers`;

GRANT USE SCHEMA ON SCHEMA dbw_proyecto_final_ehs.bronze TO `grp_data_engineers`;
GRANT SELECT ON SCHEMA dbw_proyecto_final_ehs.bronze TO `grp_data_engineers`;

GRANT USE SCHEMA ON SCHEMA dbw_proyecto_final_ehs.silver TO `grp_data_engineers`;
GRANT SELECT, MODIFY, CREATE TABLE ON SCHEMA dbw_proyecto_final_ehs.silver TO `grp_data_engineers`;

GRANT USE SCHEMA ON SCHEMA dbw_proyecto_final_ehs.golden TO `grp_data_engineers`;
GRANT SELECT, MODIFY, CREATE TABLE ON SCHEMA dbw_proyecto_final_ehs.golden TO `grp_data_engineers`;
```

## 3. Data Analysts

```sql
GRANT USE CATALOG ON CATALOG dbw_proyecto_final_ehs TO `grp_data_analysts`;
GRANT USE SCHEMA ON SCHEMA dbw_proyecto_final_ehs.golden TO `grp_data_analysts`;
GRANT SELECT ON SCHEMA dbw_proyecto_final_ehs.golden TO `grp_data_analysts`;
```

## 4. Power BI Readers

```sql
GRANT USE CATALOG ON CATALOG dbw_proyecto_final_ehs TO `grp_powerbi_readers`;
GRANT USE SCHEMA ON SCHEMA dbw_proyecto_final_ehs.golden TO `grp_powerbi_readers`;
GRANT SELECT ON SCHEMA dbw_proyecto_final_ehs.golden TO `grp_powerbi_readers`;
```

## 5. Verificación de privilegios

```sql
SHOW GRANTS ON CATALOG dbw_proyecto_final_ehs;

SHOW GRANTS ON SCHEMA dbw_proyecto_final_ehs.bronze;
SHOW GRANTS ON SCHEMA dbw_proyecto_final_ehs.silver;
SHOW GRANTS ON SCHEMA dbw_proyecto_final_ehs.golden;
```

## 6. Validación funcional recomendada

Validar con un usuario miembro de cada grupo:

- **Data Engineer:** consulta Bronze, Silver y Golden; crea/modifica tablas en Silver y Golden.
- **Data Analyst:** consulta Golden; no accede a Bronze ni Silver.
- **Power BI Reader:** consulta Golden; no accede a Bronze ni Silver.

`SHOW GRANTS` confirma los privilegios registrados. La prueba con miembros de cada grupo confirma el acceso efectivo.
